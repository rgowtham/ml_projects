```python
def matrix_dot_vector(a: list[list[int|float]], b: list[int|float]) -> list[int|float]:
    n = len(a)  # Number of rows
    m = len(a[0]) # Number of columns
    if m != len(b): # Number of columns should be equal to number of rows in vector
        return -1
    else:
        out = []
        for i in range(n):
            row = a[i]
            curr_sum = 0
            for j in range(m):
                curr_sum += (row[j] * b[j])
            out.append(curr_sum)
    return out
```

```python
print(matrix_dot_vector([[1, 2, 3, 4, 5]], [1, 1, 1, 1, 1]))
```
