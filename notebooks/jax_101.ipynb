```python
import jax.numpy as jnp
```

```python
x = jnp.arange(3.)
```

```python
def sigmoid(x):
    return jnp.sum(1 / (1 + jnp.exp(-x)))
```

```python
from jax import grad
```

```python
derivative = grad(sigmoid)
```

```python
derivative(x)
```

```python
sigmoid(x)
```

```python
eps = 1E-4
for v in jnp.eye(3):
    print((sigmoid(x + eps * v) - sigmoid(x - eps * v)) / (2 * eps))
```

So, this took me a while to understand. The function with which we compute the loss is given below,
$$
\sum\left( \sigma(x) = \frac{1}{1+e^{-x}}\right)
$$

The derivative of this method is, 
$$
\frac{d}{dx}f(x) = \sigma(x)(1 - \sigma(x))
$$

where $f(x)$ is the loss function we have. 

So, in the above notebook, we are checking if the derivative obtained from `jax.grad` produces results closer to the actual derivative value computed based on this formula
