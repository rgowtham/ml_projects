```python
import numpy as np
```

```python
class LogisticRegression:
    def __init__(self, lr=0.1, n_iter=1000):
        self.lr = lr
        self.n_iter = n_iter
        self.w = None
        self.b = None
        pass

    def _sigmoid(self, z):
        out = np.empty_like(z, dtype=float)
        pos = z >= 0
        out[pos] = 1/(1+np.exp(-z[pos]))
        out[~pos] = np.exp(z[~pos])/(1+np.exp(z[~pos]))
        return out
            
    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.w = np.zeros(n_features, dtype=float)
        self.b = 0.0 
        for _ in range(self.n_iter):
            z = X @ self.w + self.b
            y_pred = self._sigmoid(z)
            error = (y_pred - y)
            dw = X.T @ error/n_samples
            db = np.mean(error)
            self.w = self.w - (self.lr * dw)
            self.b = self.b - (self.lr * db)

    def predict(self, X_test, threshold=0.5):
        out = self.predict_proba(X_test)
        return (out >= threshold).astype(int)

    def predict_proba(self, X_test):
        z = X_test @ self.w + self.b
        proba = self._sigmoid(z)
        return proba
```

```python
model = LogisticRegression(n_iter=10)
```

```python
rng = np.random.default_rng()
```

```python
X = rng.standard_normal(size=(100, 2))
y = ((X[:,0] + (2 * X[:,1])) + rng.standard_normal(size=100) > 0).astype(int)

model.fit(X, y)

print((model.predict(X) == y).mean())
print(np.mean(model.predict_proba(X)), np.mean(y))
```
