```python
import numpy as np
from sklearn.datasets import make_blobs
from sklearn.metrics import adjusted_rand_score
```

```python
class KMeans:
    def __init__(self, n_clusters=3, n_iters=1000, distance_metric='l2'):
        self.n_clusters = n_clusters
        self.n_iters = n_iters
        self.distance_metric = distance_metric
        self.centroids = None
        self.clusters = None

    def _assign(self, X):
        distances = np.sqrt(np.sum((X[:, None, :] - self.centroids[None, :, :])**2, axis=2))
        clusters = np.argmin(distances, axis=1)
        return clusters

    def predict(self, X):
        return self._assign(X)

    def fit(self, X):
        (n_samples, n_features) = X.shape
        # self.centroids = np.random.normal(size=(self.n_clusters, n_features))
        centroids_idx = np.random.choice(n_samples, self.n_clusters, replace=False)
        self.centroids = X[centroids_idx].copy()
        for _ in range(self.n_iters):
            self.clusters = self._assign(X)
            old_centroids = self.centroids.copy()

            for c in range(self.n_clusters):
                cluster_points = X[self.clusters == c]
    
                if len(cluster_points) > 0:
                    self.centroids[c] = np.mean(cluster_points, axis=0)
        self.clusters = self._assign(X)
            if np.allclose(self.centroids, old_centroids):
                break
```

```python
model = KMeans()
```

```python
X = make_blobs(n_samples = 1000, n_features=3)
```

```python
model.fit(X[0])
```

## Using clearly separated clusters rather than `make_blobs`

```python
rng = np.random.default_rng(0)
centers = np.array([[0,0], [8, 8], [-8, -8]])
points = np.vstack([c + rng.standard_normal(size=(100, 2)) for c in centers])

model = KMeans(n_clusters=3)
model.fit(points)
print(model.centroids)
```
