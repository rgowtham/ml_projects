```python
class NeuralNet:
    def __init__(self, layer_dims=[2, 8, 1], activations=['relu', 'sigmoid'], seed=42, epochs=100, lr=0.1):
        self.layer_dims = layer_dims
        self.activations = activations
        self.epochs = epochs
        # Weights and biases
        self.w1 = None
        self.b1 = None
        self.w2 = None
        self.b2 = None

        # Intermediate layer weights
        self.a1 = None  # RELU Activation
        self.a2 = None  # Sigmoid activation
        self.z1 = None  # w1.x+b1
        self.z2 = None  # w2.a1 + b2
        self.rng = np.random.default_rng(seed)
        self.lr = lr

    def _initialize_weights(self, n_features):
        self.w1 = self.rng.standard_normal(size=(n_features, self.layer_dims[1]), dtype=float)
        self.b1 = self.rng.standard_normal(size=(n_features, 1), dtype=float)
        self.w2 = self.rng.standard_normal(size=(self.layer_dims[1], 1), dtype=float)
        self.b2 = self.rng.standard_normal(dtype=float)

    def _sigmoid(z):
        out = np.empty_like(z, dtype=float)
        pos = z >= 0
        out[pos] = 1 / (1 + np.exp(-z[pos]))
        out[~pos] = np.exp(z[~pos])/(1+np.exp(z[~pos]))
        return out
        
    def forward(self, X):
        self.z1 = X @ self.w1 + self.b1
        self.a1 = np.maximum(0, self.z1)
        self.z2 = self.a1 @ self.w2 + self.b2
        self.a2 = self._sigmoid(self.z2)
        return self.a2

    def fit(self, X, y):
        (n_samples, n_features) = X.shape
        self._initialize_weights(n_features)

        for _ in range(self.epochs):
            y_pred = self.forward(X)
            layer2_error = y_pred - y
            error_propagated = layer2_error @ self.w2.T
            layer1_error = error_propagated * (self.z1 > 0)
            
            # back propagation
            dw2 = self.a1.T @ layer2_error / n_samples
            db2 = np.mean(layer2_error, axis=0, keepdims=True)
            dw1 = X.T @ layer1_error / n_samples
            db1 = np.mean(layer1_error, axis=0, keepdims=True)

            # weight updates
            self.w1 -= (self.lr * dw1)
            self.b1 -= (self.lr * db1)
            self.w2 -= (self.lr * dw2)
            self.b2 -= (self.lr * db2)
```
